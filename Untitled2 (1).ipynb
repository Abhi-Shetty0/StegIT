```python
base_path = path + "/Lung Segmentation"

image_dir = base_path + "/CXR_png"
mask_dir = base_path + "/masks"

print("Images:", len(os.listdir(image_dir)))
print("Masks:", len(os.listdir(mask_dir)))
```

```python
import os
import cv2
import numpy as np
from torch.utils.data import Dataset

class LungDataset(Dataset):
    def __init__(self, image_dir, mask_dir):
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.images = os.listdir(image_dir)

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_name = self.images[idx]

        img_path = os.path.join(self.image_dir, img_name)

        # ✅ FIX: handle both mask naming styles
        mask_name1 = img_name.replace(".png", "_mask.png")
        mask_name2 = img_name

        mask_path1 = os.path.join(self.mask_dir, mask_name1)
        mask_path2 = os.path.join(self.mask_dir, mask_name2)

        if os.path.exists(mask_path1):
            mask_path = mask_path1
        elif os.path.exists(mask_path2):
            mask_path = mask_path2
        else:
            return self.__getitem__((idx + 1) % len(self))  # skip if not found

        image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

        # ✅ Safety check
        if image is None or mask is None:
            return self.__getitem__((idx + 1) % len(self))

        image = cv2.resize(image, (512, 512)) / 255.0
        mask = cv2.resize(mask, (512, 512)) / 255.0

        image = np.expand_dims(image, axis=0)
        mask = np.expand_dims(mask, axis=0)

        return image.astype(np.float32), mask.astype(np.float32)
```

```python
dataset = LungDataset(image_dir, mask_dir)

img, mask = dataset[0]

print(img.shape, mask.shape)
```

```python
import os

images = os.listdir(image_dir)
masks = os.listdir(mask_dir)

print("Sample images:", images[:5])
print("Sample masks:", masks[:5])
```

```python
import torch
from torch.utils.data import DataLoader

dataset = LungDataset(image_dir, mask_dir)

train_loader = DataLoader(dataset, batch_size=8, shuffle=True)
```

```python
import torch.nn as nn

class UNet(nn.Module):
    def __init__(self):
        super(UNet, self).__init__()

        def conv_block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True),
                nn.Conv2d(out_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True)
            )

        self.enc1 = conv_block(1, 64)
        self.pool1 = nn.MaxPool2d(2)

        self.enc2 = conv_block(64, 128)
        self.pool2 = nn.MaxPool2d(2)

        self.enc3 = conv_block(128, 256)
        self.pool3 = nn.MaxPool2d(2)

        self.enc4 = conv_block(256, 512)
        self.pool4 = nn.MaxPool2d(2)

        self.bottleneck = conv_block(512, 1024)

        self.up4 = nn.ConvTranspose2d(1024, 512, 2, stride=2)
        self.dec4 = conv_block(1024, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.dec3 = conv_block(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.dec2 = conv_block(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.dec1 = conv_block(128, 64)

        self.final = nn.Conv2d(64, 1, kernel_size=1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool1(e1))
        e3 = self.enc3(self.pool2(e2))
        e4 = self.enc4(self.pool3(e3))

        b = self.bottleneck(self.pool4(e4))

        d4 = self.up4(b)
        d4 = torch.cat([d4, e4], dim=1)
        d4 = self.dec4(d4)

        d3 = self.up3(d4)
        d3 = torch.cat([d3, e3], dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)
        d1 = self.dec1(d1)

        return torch.sigmoid(self.final(d1))
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = UNet().to(device)

criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
from tqdm import tqdm
import torch # Assuming torch is already imported

epochs = 5

for epoch in range(epochs):
    model.train()
    total_loss = 0

    # 1. Wrap the train_loader with tqdm
    progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}")

    for images, masks in progress_bar:
        # Best practice: Assuming DataLoader yields tensors.
        # If they are numpy arrays, use torch.as_tensor(images).to(device)
        images = images.to(device)
        masks = masks.to(device)

        # Forward pass
        preds = model(images)
        loss = criterion(preds, masks)

        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

        # 2. Update the progress bar with the current batch loss
        progress_bar.set_postfix(loss=f"{loss.item():.4f}")

    # Calculate and print the average loss for the entire epoch
    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch+1} Completed | Average Loss: {avg_loss:.4f}\n")
```

```python
import matplotlib.pyplot as plt

model.eval()

img, mask = dataset[0]

with torch.no_grad():
    pred = model(torch.tensor(img).unsqueeze(0).to(device)).cpu().numpy()[0][0]

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Input")
plt.imshow(img[0], cmap='gray')

plt.subplot(1,3,2)
plt.title("Ground Truth")
plt.imshow(mask[0], cmap='gray')

plt.subplot(1,3,3)
plt.title("Prediction")
plt.imshow(pred, cmap='gray')

plt.show()
```

```python


# Normalize
img = img / 255.0

# Add channel + batch dimension → (1,1,H,W)
img_tensor = torch.tensor(img, dtype=torch.float32).unsqueeze(0).unsqueeze(0).to(device)

# Model prediction
model.eval()
with torch.no_grad():
    pred = model(img_tensor)
    pred = torch.sigmoid(pred)
    pred = pred.cpu().numpy()[0][0]

# Convert to binary mask
pred_binary = (pred > 0.5).astype(np.uint8)

# Plot results
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.title("Input Image")
plt.imshow(img, cmap='gray')

plt.subplot(1,3,2)
plt.title("Predicted Mask")
plt.imshow(pred, cmap='gray')

plt.subplot(1,3,3)
plt.title("Binary Mask")  
plt.imshow(pred_binary, cmap='gray')

plt.show()
```
